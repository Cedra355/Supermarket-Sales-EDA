# PART 3 - FEATURE ANALYSIS

```python
# Numerical feature analysis
numerical_cols = df.select_dtypes(include="number").columns

# Descriptive statistics
numerical_summary = df[numerical_cols].describe().T

# Add mean, median, variance and skewness
numerical_summary["mean"] = df[numerical_cols].mean()
numerical_summary["median"] = df[numerical_cols].median()
numerical_summary["variance"] = df[numerical_cols].var()
numerical_summary["skewness"] = df[numerical_cols].skew()

display(numerical_summary)
```

```python
# Calculate key statistics for numerical features
feature_stats = df[numerical_cols].agg(["min", "max", "mean", "median"]).T

feature_stats.columns = ["Minimum", "Maximum", "Average", "Median"]

display(feature_stats.round(2))
```

```python
# Categorial feature analysis
categorical_cols = df.select_dtypes(include=["object", "category"]).columns

# Display category counts and percentages
for col in categorical_cols:
  print(f"\n--- {col} ---")

  counts = df[col].value_counts(dropna=False)
  percentages = (
      df[col].value_counts(normalize=True, dropna=False) * 100
  )

  category_summary = pd.DataFrame({
      "Count": counts,
      "Percentage": percentages.round(2)
  })

  display(category_summary.round(2))
```

```python
# Feature distributions

import matplotlib.pyplot as plt
import seaborn as sns

# Plot distributions of numerical features
df[numerical_cols].hist(
figsize=(16, 12),
bins = 20,
edgecolor = "black"
)

plt.suptitle("Distribution of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
# Visualize numerical features to inspect outliers
for col in numerical_cols:
  plt.figure(figsize=(8, 3))

  sns.boxplot(x = df[col])
  plt.title(f"Boxplot of {col}")

  plt.tight_layout()
  plt.show()
```

```python
# Detect potential outliers using the IQR method
outlier_results = []

for col in numerical_cols:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1

  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR

  outlier_count = (
   (df[col] < lower_bound) | (df[col] > upper_bound)
  ).sum()

  outlier_results.append({
      "Feature": col,
      "Lower Bound": lower_bound,
      "Upper Bound": upper_bound,
      "Outlier Count": outlier_count,
      "Outlier Percentage (%)": (
          outlier_count / len(df) * 100
      )
      })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary.round(2))
```

```python
# Compare revenue and transaction stastistics by category
business_features = [
    "Product line",
    "Branch",
    "Customer type",
    "Payment"
]

for col in business_features:
  if col in df.columns:
    print(f"\nRevenue Analysis for {col}:")

    summary = df.groupby(col, dropna = False)["Total"].agg(
        Transactions = ("Total", "siza"),
        Total_Revenue = ("Total", "sum"),
        Average_Sale = ("Total", "mean"),
        Median_Sale = ("Total", "median")
    ).sort_values(
        "Total_Revenue", ascending = False
    )

    display(summary.round(2))
```

# PART 4 - CORRELATION ANALYSIS

```python
# Select suitable numerical features for correlation analysis
numerical_features = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

# Calculate the correlation matrix
correlation_matrix = df[numerical_features].corr()

display(correlation_matrix.round(2))
```

```python
# Create a heatmap to visualize correlations
plt.figure(figsize = (8, 6))

sns.heatmap(
    correlation_matrix,
    annot = True,
    cmap="coolwarm",
    fmt = ".2f",
    linewidths = 0.5,
    vmin = -1,
    vmax = 1
)

plt.title("Correlation Heatmap of Sales Features")
plt.tight_layout()
plt.show()
```

```python
# Extract unique pairs of features and their correlations
correlation_pairs = correlation_matrix.where(
    ~correlation_matrix.isna()
).stack().reset_index()

correlation_pairs.columns = [
    "Feature 1", "Feature 2", "Correlation"
]

# Remove duplicate pairs and self-correlations
correlation_pairs = correlation_pairs[
    correlation_pairs["Feature 1"] < correlation_pairs["Feature 2"]
].copy()

# Classify the strength of each relationship
def classify_correlation(value):
    strength = abs(value)

    if strength >= 0.70:
        return "Strong"
    elif strength >= 0.40:
        return "Moderate"
    else:
        return "Weak"

correlation_pairs["Strength"] = (
    correlation_pairs["Correlation"].apply(classify_correlation)
)

# Identify the direction of each relationship
correlation_pairs["Direction"] = correlation_pairs[
    "Correlation"
].apply(
    lambda x: "Positive" if x > 0
    else "Negative" if x < 0
    else "No linear direction"
)

# Display relationships from strongest to weakest
display(
    correlation_pairs.sort_values(
        "Correlation",
        key=lambda x: x.abs(),
        ascending=False
    ).round(2)
)
```

## Correlation Analysis - Findings

The correlation analysis shows no strong relationships between Sales, Quantity, Discount, and Profit.

- **Sales and Profit:** Sales and Profit have a moderate positive correlation of **0.48**. This means that higher sales are generally linked with higher profit.

- **Discount and Profit:** Discount and Profit have a weak negative correlation of **-0.22**. Higher discounts are slightly linked with lower profit.

- **Sales and Quantity:** Sales and Quantity have a weak positive correlation of **0.20**. Orders with more items may have slightly higher sales.

- **Weak relationships:** Sales and Discount have a correlation of **-0.03**, while Quantity and Discount have a correlation of **0.01**. These values are very close to zero, so they have almost no linear relationship.

- **Business insight:** Discounts should be monitored because higher discounts may reduce profit. However, correlation does not prove that discounts directly cause lower profit.

# PART 6 - BUSINESS INSIGHTS

The dataset has 9,994 order-line records from 5,009 unique orders. The following insights are based on the results from the analysis.

1. **Overall Performance:** The business made sales of **$2,297,200.86** and a total profit of **286,397.02** from 9.994 order-lines.

2. **Technology Leads Sales and Profit:** Technology was the best-performing category. It generated **$836,154.03** in sales and **145,454.95** in profit, which were the highest among the three categories.

3. **Office Supplies Contributes Substantial Profit:** Office Supplies generated **$719,047.03** in sales and made **122,490.80** in profit. This shows that it was also an important category for the business.

4. **Furniture has a Comparatively Low Profit Margin:** Furniture generated high sales of **$741,999.80**, but its profit was only **18,451.27**. Its profit margin was around **2.5%**, which is much lower than the other categories.

5. **Consumers Account for the Largest Sales Share:** Consumer customers generated **$1,161,401.34** in sales, which is about **50.6%** of the company’s total sales.

6. **Home Office has the Highest Segment Profit Margin:** The Home Office segment made **$60,298.68** profit from **429,653.15** in sales. Its profit margin was around **14%**, which was higher than the Consumer segment’s margin of about **11.5%**.

7. **The West is the Top Region by Sales and Profit:** The West region performed best overall, generating **$725,457.82** in sales and **108,418.45** in profit.

8. **Central has the Lowest Regional Profit Margin:** Central generated **$501,239.89** in sales and **39,706.36** in profit. Its profit margin was around **7.9%**, and it also had the highest average discount of **24%**.

9. **Texas Combines High Sales with the Largest State-level Loss:** Texas made **$170,188.05** in sales but had a loss of **25,729.36**. This shows that high sales do not always result in good profit.

10. **Tables is the Largest Loss-making Sub-category:** Tables generated **$206,965.53** in sales but made a loss of **17,725.48**.

11. **Bookcases and Supplies also Recorded Losses:** Bookcases made a loss of **$3,472.56**, while Supplies made a loss of **1,189.10**.

12. **Copiers had the Highest Sub-category Profit:** Copiers made the highest profit of all sub-categories, earning **$55,617.82** profit from **149,528.03** in sales.

13. **Phones and Chairs Lead Sub-category Sales:** Phones had the highest sales of **$330,007.05**, while Chairs were close behind with **328,449.10** in sales.

14. **Loss-making order lines are a significant minority:** There were **1,871** loss-making order lines, which is **18.7%** of all records. These losses added up to **$156,131.29**.

15. **No-discount Lines had Higher Average Profit than 20%-Discount Lines:** Orders without a discount had an average profit of **$66.90**. Orders with a 20% discount had a lower average profit of **24.70**.

16. **The 30% Discount Group frequently Recorded Losses:** Out of 227 order lines with a 30% discount, **91.6%** made a loss. Their average profit was **−$45.68**.

17. **Fourth-quarter Sales were Highest:** Sales were highest in the fourth quarter at **$368,948.62**. First-quarter sales were much lower at **116,661.29**.

18. **Sales Rose while Profit Fell from 2013 to 2014:** Sales increased from **$261,514.21** in 2013 to **280,923.61** in 2014. However, profit decreased from **43,667.09** to **27,952.92**, so the increase in sales did not lead to higher profit.

# PART 7 - BUSINESS IMPROVEMENT RECOMMENDATIONS

1. **Review the discount policy:** Discounts of 30% or more should be reviewed because most of these orders made a loss. The business can set discount limits or approve large discounts only for selected products.

2. **Improve Furniture profitability:** Furniture had high sales but a very low profit margin. The company should check product costs, shipping costs, and discounts in this category to improve profit.

3. **Focus on loss-making sub-categories:** Tables, Bookcases, and Supplies made losses. The company should review their pricing and costs before continuing the same sales strategy.

4. **Investigate losses in Texas and Central:** Texas had the highest state-level loss, while Central had the lowest regional profit margin. The business should review discounting, product mix, and delivery costs in these areas.

5. **Promote profitable products:** Copiers, Phones, Accessories, and Paper generated strong profits. These products can be promoted through marketing campaigns, bundles, or recommendations to customers.

6. **Target high-value customer segments:** Consumer customers generated the highest sales, while Home Office customers had the highest profit margin. Different offers can be created for both groups to increase revenue and profit.

7. **Prepare for fourth-quarter demand:** Sales were highest in the fourth quarter. The company should plan inventory, staff, and promotions early to avoid stock shortages during this period.
